TRAIN/TEST SPLIT

```python
x = df['Text']
```

```python
y = df['Sentiment']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("X_train",X_train.shape)
print("X_test",X_test.shape)
print("Y_train",Y_train.shape)
print("Y_test",Y_test.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)
```

```python
X_test_tfidf = tfidf.transform(X_test)
```

```python
print("Training : ",X_train_tfidf.shape)
print("Testing : ",X_test_tfidf.shape)
```

USING LOGISTIC REGRESSION TO TRAIN THE MODEL

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, Y_train)
```

PREDICTING and CHECKING THE MODELS ACCURACY

```python
y_pred = model.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test,y_pred)

print("Accuracy = ", accuracy)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print("Confusion Matrix: ",confusion_matrix(Y_test,y_pred))

print("Classification Report: ",classification_report(Y_test,y_pred))
```

TESTING THE MODEL WITH NEW REVIEWS

```python
new_reviews = [
    "This phone is amazing and the battery lasts all day.",
    "Very bad product, it stopped working after two days.",
    "This product is okay, nothing special."
]

new_reviews_tfidf = tfidf.transform(new_reviews)

prediction = model.predict(new_reviews_tfidf)

for review, prediction in zip(new_reviews, prediction):
    print("Review: ", review)
    print("Predicted Sentiment: ", prediction)
    print()
```

CREATING PIPELINE

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('tfidf',TfidfVectorizer()),
    ('model', LogisticRegression(max_iter=1000))
])
```

```python
pipeline.fit(X_train, Y_train)
```

```python
y_pred_pipeline = pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

pipeline_accuracy = accuracy_score(Y_test, y_pred_pipeline)

print("Pipeline Accuracy: ", pipeline_accuracy)
```

```python
import joblib

joblib.dump(pipeline, 'sentiment_pipeline.pkl')
```
